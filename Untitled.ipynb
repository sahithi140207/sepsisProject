```python
# Combine the extracted ICU measurements
frames = [vital_signs, extra_vitals]

if not spo2_data.empty:
    frames.append(spo2_data)

combined = pd.concat(frames, ignore_index=True)

combined["charttime"] = pd.to_datetime(
    combined["charttime"], errors="coerce"
)

combined = combined.dropna(
    subset=["stay_id", "charttime", "valuenum"]
)

combined["hour"] = combined["charttime"].dt.floor("h")

# Median measurement per stay, hour, and item
hourly_all = (
    combined.groupby(["stay_id", "hour", "itemid"])["valuenum"]
    .median()
    .unstack("itemid")
    .reset_index()
)

# Rename the columns
hourly_all = hourly_all.rename(columns={
    220045: "heart_rate_bpm",
    220210: "respiratory_rate",
    223761: "temperature_fahrenheit",
    223762: "temperature_celsius",
    220050: "arterial_systolic",
    220051: "arterial_diastolic",
    220052: "arterial_map",
    220179: "noninvasive_systolic",
    220180: "noninvasive_diastolic",
    220181: "noninvasive_map",
    220277: "spo2_percent"
})

# Combine temperature readings in Celsius
if "temperature_fahrenheit" in hourly_all.columns:
    if "temperature_celsius" not in hourly_all.columns:
        hourly_all["temperature_celsius"] = np.nan

    missing_temp = (
        hourly_all["temperature_celsius"].isna()
        & hourly_all["temperature_fahrenheit"].notna()
    )

    hourly_all.loc[missing_temp, "temperature_celsius"] = (
        hourly_all.loc[missing_temp, "temperature_fahrenheit"] - 32
    ) * 5 / 9

# Prefer non-invasive blood pressure when available,
# otherwise use arterial blood pressure.
for output_col, noninvasive_col, arterial_col in [
    ("systolic_bp", "noninvasive_systolic", "arterial_systolic"),
    ("diastolic_bp", "noninvasive_diastolic", "arterial_diastolic"),
    ("mean_bp", "noninvasive_map", "arterial_map")
]:
    hourly_all[output_col] = hourly_all[noninvasive_col].combine_first(
        hourly_all[arterial_col]
    )

hourly_all = hourly_all.sort_values(
    ["stay_id", "hour"]
).reset_index(drop=True)

print("Hourly dataset shape:", hourly_all.shape)
print("ICU stays:", hourly_all["stay_id"].nunique())
display(hourly_all.head())

display(hourly_all.isna().sum().to_frame("missing_rows"))
```

```python
ranges = {
    "heart_rate_bpm": (20, 250),
    "respiratory_rate": (4, 80),
    "temperature_celsius": (25, 45),
    "systolic_bp": (40, 300),
    "diastolic_bp": (10, 200),
    "mean_bp": (15, 250),
    "spo2_percent": (50, 100)
}

hourly_model = hourly_all.copy()

for col, (low, high) in ranges.items():
    if col in hourly_model.columns:
        invalid = ~hourly_model[col].between(low, high)
        invalid_count = int(invalid.sum())
        hourly_model.loc[invalid, col] = np.nan
        print(f"{col}: set {invalid_count} out-of-range values to missing")

print("\nCleaned dataset:")
display(hourly_model.head())
```

```python
# Add the ICU stay ID to lactate results using admission and time
lactate_data["charttime"] = pd.to_datetime(
    lactate_data["charttime"], errors="coerce"
)

stays_for_merge = icu_stays[
    ["subject_id", "hadm_id", "stay_id", "intime", "outtime"]
].copy()

stays_for_merge["hadm_id"] = pd.to_numeric(
    stays_for_merge["hadm_id"], errors="coerce"
)
lactate_data["hadm_id"] = pd.to_numeric(
    lactate_data["hadm_id"], errors="coerce"
)

stays_for_merge["intime"] = pd.to_datetime(stays_for_merge["intime"])
stays_for_merge["outtime"] = pd.to_datetime(stays_for_merge["outtime"])

lactate_matched = lactate_data.merge(
    stays_for_merge,
    on=["subject_id", "hadm_id"],
    how="inner"
)

lactate_matched = lactate_matched[
    lactate_matched["charttime"].between(
        lactate_matched["intime"],
        lactate_matched["outtime"]
    )
].copy()

lactate_matched["hour"] = lactate_matched["charttime"].dt.floor("h")

lactate_hourly = (
    lactate_matched.groupby(["stay_id", "hour"])["valuenum"]
    .median()
    .reset_index(name="lactate")
)

print("Lactate results matched to ICU stays:", len(lactate_matched))
print("ICU stays with matched lactate:", lactate_matched["stay_id"].nunique())

display(lactate_hourly.head())
```

```python
# Check how many usable measurements remain
features = [
    "heart_rate_bpm",
    "respiratory_rate",
    "temperature_celsius",
    "systolic_bp",
    "diastolic_bp",
    "mean_bp",
    "spo2_percent",
    "lactate"
]

display(hourly_model[features].describe().T)

print("\nMissing percentage:")
display(
    (hourly_model[features].isna().mean() * 100)
    .round(1)
    .to_frame("missing_percent")
)
```

```python
state_features = [
    "heart_rate_bpm",
    "respiratory_rate",
    "temperature_celsius",
    "systolic_bp",
    "diastolic_bp",
    "mean_bp",
    "spo2_percent"
]

states = hourly_model[
    ["stay_id", "hour"] + state_features
].copy()

states = states.sort_values(["stay_id", "hour"]).reset_index(drop=True)

# Impute missing values within each ICU stay, then use global medians
for col in state_features:
    states[col] = states.groupby("stay_id")[col].transform(
        lambda series: series.fillna(series.median())
    )
    states[col] = states[col].fillna(states[col].median())

print("State dataset shape:", states.shape)
print("Remaining missing feature values:", int(states[state_features].isna().sum().sum()))

display(states.head())
```

```python
# Simple research labels based on vital-sign patterns.
# These are NOT validated clinical thresholds or treatment recommendations.

states["simulated_action"] = 0

review_pattern = (
    (states["heart_rate_bpm"] > 110)
    | (states["respiratory_rate"] > 24)
    | (states["systolic_bp"] < 90)
    | (states["spo2_percent"] < 92)
)

states.loc[review_pattern, "simulated_action"] = 1

stronger_review_pattern = (
    (states["heart_rate_bpm"] > 130)
    | (states["respiratory_rate"] > 30)
    | (states["systolic_bp"] < 80)
    | (states["spo2_percent"] < 88)
)

states.loc[stronger_review_pattern, "simulated_action"] = 2

display(states["simulated_action"].value_counts().sort_index().to_frame("hours"))
```

```python
# Sort by patient stay and time
states = states.sort_values(
    ["stay_id", "hour"]
).reset_index(drop=True)

# Next-hour values within the same ICU stay
next_states = states.groupby("stay_id")[state_features].shift(-1)

# Keep transitions only when a next hour exists
has_next = states.groupby("stay_id")["hour"].shift(-1).notna()

transitions = states.loc[
    has_next, ["stay_id", "hour", "simulated_action"] + state_features
].copy()

transitions["next_hour"] = (
    states.groupby("stay_id")["hour"].shift(-1)[has_next].values
)

for col in state_features:
    transitions["next_" + col] = next_states.loc[has_next, col].values

print("Number of transitions:", len(transitions))
print("ICU stays:", transitions["stay_id"].nunique())

display(transitions.head())
```

```python
# A toy research reward based on next-hour vital-sign patterns.
# These thresholds are illustrative only, not clinical recommendations.

def pattern_score(df, prefix=""):
    hr = df[prefix + "heart_rate_bpm"]
    rr = df[prefix + "respiratory_rate"]
    sbp = df[prefix + "systolic_bp"]
    spo2 = df[prefix + "spo2_percent"]

    score = (
        (hr > 110).astype(int)
        + (rr > 24).astype(int)
        + (sbp < 90).astype(int)
        + (spo2 < 92).astype(int)
    )
    return score

transitions["current_pattern_score"] = pattern_score(transitions)
transitions["next_pattern_score"] = pattern_score(
    transitions, prefix="next_"
)

# Positive reward means the illustrative pattern score fell;
# negative reward means it rose.
transitions["reward"] = (
    transitions["current_pattern_score"]
    - transitions["next_pattern_score"]
)

display(transitions[[
    "stay_id", "hour", "simulated_action",
    "current_pattern_score", "next_pattern_score", "reward"
]].head(10))

print("\nReward counts:")
display(transitions["reward"].value_counts().sort_index().to_frame("transitions"))
```

```python
# Step 29: Prepare data for offline RL

import pandas as pd
import numpy as np
from sklearn.model_selection import GroupShuffleSplit

# Features describing the current patient state
state_features = [
    "heart_rate_bpm",
    "respiratory_rate",
    "temperature_celsius",
    "systolic_bp",
    "diastolic_bp",
    "mean_bp",
    "spo2_percent"
]

# Separate current states, actions, rewards and next states
X = transitions[state_features].copy()
actions = transitions["simulated_action"].copy()
rewards = transitions["reward"].copy()
next_X = transitions[["next_" + col for col in state_features]].copy()
groups = transitions["stay_id"]

# Split by ICU stay to reduce patient-level data leakage
splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.2,
    random_state=42
)

train_idx, test_idx = next(splitter.split(X, rewards, groups))

print("Training transitions:", len(train_idx))
print("Testing transitions:", len(test_idx))
print("Training ICU stays:", groups.iloc[train_idx].nunique())
print("Testing ICU stays:", groups.iloc[test_idx].nunique())
```

```python
# Step 30: Train a baseline model

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# Get training and testing data
X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_train = actions.iloc[train_idx]
y_test = actions.iloc[test_idx]

# Train the model
baseline_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight="balanced"
)

baseline_model.fit(X_train, y_train)

# Predict actions for the test data
baseline_predictions = baseline_model.predict(X_test)

# Evaluate predictions
print("Baseline Action Prediction Accuracy:",
      round(accuracy_score(y_test, baseline_predictions), 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    baseline_predictions,
    labels=[0, 1, 2],
    zero_division=0
))
```

```python
# Step 31: Feature importance

import pandas as pd
import matplotlib.pyplot as plt

importance = pd.Series(
    baseline_model.feature_importances_,
    index=state_features
).sort_values(ascending=True)

print("Feature importance:")
display(importance.sort_values(ascending=False).to_frame("Importance"))

importance.plot(kind="barh", figsize=(8, 5))
plt.title("Baseline Model: Vital-Sign Feature Importance")
plt.xlabel("Importance")
plt.ylabel("Vital Sign")
plt.tight_layout()
plt.show()
```

```python
# Step 32: Learn a reward model from offline transitions

from sklearn.ensemble import RandomForestRegressor
import numpy as np
import pandas as pd

# Use only training transitions
X_train = X.iloc[train_idx]
y_reward = rewards.iloc[train_idx]
a_train = actions.iloc[train_idx]

# Create a separate reward model for each simulated action
reward_models = {}

for action in [0, 1, 2]:
    mask = (a_train == action)

    model = RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        min_samples_leaf=5
    )

    model.fit(X_train.loc[mask], y_reward.loc[mask])
    reward_models[action] = model

print("Reward models trained for actions 0, 1 and 2.")
```

```python
# Step 33: Select the action with the highest predicted reward

X_test = X.iloc[test_idx]

predicted_rewards = np.column_stack([
    reward_models[action].predict(X_test)
    for action in [0, 1, 2]
])

policy_actions = predicted_rewards.argmax(axis=1)

policy_results = pd.DataFrame({
    "stay_id": transitions.iloc[test_idx]["stay_id"].to_numpy(),
    "hour": transitions.iloc[test_idx]["hour"].to_numpy(),
    "recorded_simulated_action": actions.iloc[test_idx].to_numpy(),
    "policy_action": policy_actions,
    "predicted_reward": predicted_rewards.max(axis=1)
})

print("Policy action distribution:")
display(
    policy_results["policy_action"]
    .value_counts()
    .sort_index()
    .rename_axis("Action")
    .to_frame("Count")
)

display(policy_results.head(10))
```

```python
# Step 34: Compare actions

import matplotlib.pyplot as plt
import pandas as pd

action_comparison = pd.DataFrame({
    "Recorded simulated actions":
        policy_results["recorded_simulated_action"].value_counts(),
    "Learned policy actions":
        policy_results["policy_action"].value_counts()
}).fillna(0).sort_index()

action_comparison.index.name = "Action"
display(action_comparison)

action_comparison.plot(kind="bar", figsize=(8, 5))
plt.title("Recorded Actions vs Learned Policy Actions")
plt.xlabel("Action (0 = routine, 1 = review, 2 = stronger review)")
plt.ylabel("Number of Transitions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
